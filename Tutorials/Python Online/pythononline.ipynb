```python
#Example 134

import matplotlib.pyplot as plt

from scipy import stats

x = [5,7,8,3,18,2,1,10,5,12,14,9]

y = [98, 87, 88, 86,109,99,120, 130, 150, 200, 250, 300]

slope, intercept, r, p, std_err = stats.linregress(x,y)

def myfunc (x):

    return slope*x + intercept

mymodel = list(map(myfunc, x))

plt.scatter(x,y)

plt.plot(x, mymodel)

plt.show()
```

```python
# Example 135

print("slope is :", slope)

print("relationship : ", r)

print("prediction is : ", p)

print("standard errors is :", std_err)
```

```python
#Example 139

import numpy

from sklearn import linear_model

X = numpy.array([3.48, 2.45,2.11,0.17, 1.90, 1.56,4.92,4.38, 4.98, 4.53, 3.79,5.88]).reshape(-1,1)

no = 0

yes = 1

y = numpy.array([no,no,no,no,no,no,yes,yes,yes,yes,yes,yes])

logr = linear_model.LogisticRegression() 

logr.fit(X,y)

pred = logr.predict(numpy.array([3.46]).reshape(-1,1))

print(pred)
```
